# Exercise 1

This exercise shows how SymPy permits to calculate the velocity addition in special relativity. If a reference frame $R_2$ moves at a speed v in $R_1$, a particle of speed $u_1$ has a speed in $R_2$ given by:

$$
\begin{align*}
U_2 &=\frac{ u_1-v} {1 - \frac{u_1 v} {c^2}}
\end{align*}
$$

We will demonstrate this result by using SymPy by calculating with the Lorentz transformation this term:

$$
\begin{align*}
\frac{dx_2} {dt_2} &= \frac{dx_2} {dt_1} \frac{dt_1} {dt_2}
\end{align*}
$$

Recall that the Lorentz transformation applied to a particle of trajectory $t_1 \rightarrow x_1(t_1)$ or $t_2 \rightarrow x_2(t_2)$ is:

$$
\begin{align*}
\gamma &= \frac{1} {\sqrt {1 - \frac{v^2} {c^2}} } \\
X_2(t_2) &= \gamma \left( x_1(t_1) - v t_1 \right) \\
T_2 &= \gamma \left( t_1 - v x_1(t_1) \right) \\
\end{align*}
$$


* Declare the variables v,c, $t_1,t_2$ as real
* Declare $x_1, x_2$ as functions (We indicate SymPy that those symbols depend on $t_1$ or $t_2$)
* Declare the Lorentz transformation between $(x_1,t_1)$ and $(x_2,t_2)$ as an equation (Hint: sym.Eq)

```python
v,c = sym.symbols("v c")
gamma = 1 / sym.sqrt(1 - (v/c)**2) 
t1,t2 = sym.symbols ("t_1 t_2")
x1, x2 = sym.symbols("x_1 x_2", cls=sym.Function)
eqs  = [ sym.Eq(  x2(t2), gamma * ( x1(t1) - v * t1 ) ), 
        sym.Eq(t2, gamma * ( t1 - v * x1(t1) / c**2 ))]
eqs[0]
```

```python
eqs[1]
```

* Use the solve function to create a map for the Lorentz transformation

(Hint: solve by targeting $x_2(t_2)$ and $t_2$)

```python
eqs_sol = sym.solve(eqs, (x2(t2),t2))
eqs_sol
```

* Calculate the inverse Lorentz transformation as before (Hint: solve with $x_2(t_2)$ and $t_2$) as targets
* Create an equation with those solutions

```python
inv_eqs_sol = sym.solve( eqs, (x1(t1), t1))
inv_eqs = [ sym.Eq( x1(t1), inv_eqs_sol[x1(t1)]), sym.Eq( t1, inv_eqs_sol[t1])]
inv_eqs[0].simplify()
```

```python
inv_eqs[1].simplify()
```

We will now calculate the particle speed relatively to $R_2$ as:

\begin{align*}
\frac {dx_2} {dt_2} &= \frac {dx_2} {dt_1} \frac {dt_1} {dt_2}
\end{align*}

* Calculate $\frac{dx_2} {dt_1}$ using the first direct Lorentz equation
* Calculate $\frac{dt_1} {dt_2}$ using the second equation of inverse Lorentz transformation

```python
# dx2 / dt2 = dx2 / dt1  *  dt1 / dt2


dx2_dt1 = sym.diff ( eqs_sol[x2(t2)], t1)
dx2_dt1
```

```python
dt1_dt2 = sym.diff ( inv_eqs_sol[t1], t2)
dt1_dt2
```

* Deduce $\frac{dx_2} {dt_2}$ from the previous calculation and build a final equation with this derivation
* Declare the $u_1, u_2$ particle speed in the $R_1$ and $R_2$ reference frame.
* Substitute in the final equation these last speeds to their formal forms $\frac{d} {dt_*}x_*(t_*)$.

```python
# dx2 / dt2 = dx2 / dt1  *  dt1 / dt2
final_eq_1 = sym.Eq( sym.diff ( x2(t2), t2) , dx2_dt1 * dt1_dt2)
final_eq_1.simplify()
```

```python
u1,u2 = sym.symbols("u_1 u_2")
final_eq_2 = final_eq_1.subs( {sym.diff ( x1(t1)):u1, sym.diff ( x2(t2)):u2 } )
final_eq_2.simplify()
```

* Solve the final equation by targeting the $u_2$ speed and simplify the result
* Compare to the previous theoretical formula for the speed addition.

```python
final_sol = sym.solve( final_eq_2, u2)
sol_eq = sym.Eq( u2, final_sol[0].simplify())
sol_eq
```

## Part 3: Introducing the proper time and acceleration

We will find how the quantity $\frac{d^2 x_1} {d\tau^2}$ ($\tau$ is the proper time of the particle) is transformed between $R_1$ and $R_2$.

The proper time of a particle is the time that is measured by a clock that follows the movement of the particle. At each moment t_2, the derivative $\frac{dt_2} {d\tau}$ is calculated using a proper reference frame $R_1(t_2)$ (different at each time) centered at the particle immobile with respect to $R_1$ although with a speed of $v(t_2)$ with respect to $R_2$. Consequently, the time $\tau$ and $t_1$ at the origin instant have the smae derivative with respect to t_2 (which is not always the case later on since the particle might have accelerated). We have the following formulae coming from the Lorentz transformation, especially (5) that gives the time $t_{f,2}$ of a particle given its proper time $\tau_f$:

$$
\begin{align*}
\gamma &= \frac{1} {\sqrt {1 - \frac{v^2} {c^2}} } \\
\frac{dt_2} {d\tau} &= \frac{dt_2} {dt_1} = \gamma \\
T_f &= \int _0 ^{\tau_f} { \frac{1} {\sqrt {1 - \frac{v(\tau)^2} {c^2}} } d\tau} \ (5)\\
\tau_f &= \int _0 ^{t_f} {\sqrt {1 - \frac{v(t_2)^2} {c^2}} dt_2} \ (6)\\
\end{align*}
$$

* Justify from equation (6) from the previous equations:
* In case the particle is constantly accelerated at g $m/s^2$ with respect to $R_2$, calculate the proper time $\tau_f$ given the time $t_f$ in $R_2$ by applying (6)
