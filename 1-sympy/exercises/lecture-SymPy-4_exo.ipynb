## Differential linear system of equations

Given $X_0 = \begin{bmatrix} x_0(0)\\ x_1(0)\\ x_2(0)\end{bmatrix}$, 
We will solve the following 3D temporal differential linear system of equations:

$$
\begin{align*}
  X_0'(t) &= 2x_0(t) - 2x_1(t) + 2x_2(t)\\
  X_1'(t) &=  3x_1(t) - x_2(t) \\
  X_2'(t) &= 2x_0(t) + x_1(t) + 3x_2(t)\\
\end{align*}
$$

We set $X(t) = \begin{bmatrix} x_0(t)\\ x_1(t)\\ x_2(t)\end{bmatrix}$

We will show that $X_s(t) = e^{At}X(0)$ is a solution of the system.


___
* Declare the variables and the symbols corresponding the previous equations.
* Declare the A matrix corresponding to the problem.

```python
import sympy as sym

# Declare the time variable
t = sym.symbols("t")

# Declare the unknown functions
x0 = sym.Function("x0")(t)
x1 = sym.Function("x1")(t)
x2 = sym.Function("x2")(t)

# Define the state vector X(t)
X = sym.Matrix([x0, x1, x2])

# Display the vector
print("X(t) =")
sym.pprint(X)
```

```python
# Define the coefficient matrix A
A = sym.Matrix([
    [2, -2,  2],
    [0,  3, -1],
    [2,  1,  3]
])

print("Matrix A =")
sym.pprint(A)
```

```python
# Verify the differential system
X_prime = X.diff(t)
system_rhs = A * X

print("X'(t) =")
sym.pprint(X_prime)

print("A * X(t) =")
sym.pprint(system_rhs)
```

* Set the Xst vector to the solution of the system $X_s(t) = e^{At} X(0)$

```python
# Declare the initial conditions
x00, x10, x20 = sym.symbols("x_0(0) x_1(0) x_2(0)")

X0 = sym.Matrix([x00, x10, x20])

# Calculate the solution Xs(t) = exp(A*t) * X(0)
Xst = (A * t).exp() * X0

print("Xs(t) =")
sym.pprint(Xst)
```

* Show that Xst is a solution of the system, ie:
  * $X_s(0) = X_0$
  * $X'_s(t) = A X_s(t)$
    

```python
# 1. Verify that Xs(0) = X0
Xs_0 = Xst.subs(t, 0)

print("Xs(0) =")
sym.pprint(Xs_0)

print("Xs(0) == X0:", sym.simplify(Xs_0 - X0) == sym.zeros(3, 1))
```

```python
# 2. Verify that Xs'(t) = A * Xs(t)
Xst_prime = Xst.diff(t)
AXst = A * Xst

verification = (Xst_prime - AXst).applyfunc(sym.simplify)

print("Xs'(t) - A*Xs(t) =")
sym.pprint(verification)

print("Xs'(t) == A*Xs(t):", verification == sym.zeros(3, 1))
```
