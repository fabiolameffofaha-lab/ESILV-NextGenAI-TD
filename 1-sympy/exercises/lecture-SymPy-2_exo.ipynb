## Exercise 2.3

A continued fraction is similar to this form:

$a_0 + \frac{1} {a_1 + \frac{1} { a_2 + \frac{1} {a_3 + \frac{1}{a_4 + ... }} } }$.

* Code a recursive function continued_fraction that creates this fraction given a finite list of $[a_0,a_1..,a_n]$. Observe that the last term has the pattern $...1/a_n$
* Test it with 10 a_n symbols (symbols("a:10"))
  

```python
def continued_fraction ( ans: List[object]) -> object:
    if len(ans) == 1:
        return ans[0]
    else:
        return ans[0] + 1 / ( continued_fraction( ans[1:]))

n = 10
ans = sym.symbols("a:"+ str (n))
continued_fraction(ans)
```

* Apply cancel function to flat the continued fraction of 7 $a_n$ symbols.
* * What kind of different terms appear in the numerator and the denominator?

```python
n = 7
ans = sym.symbols("a:"+ str (n))
sym.cancel(continued_fraction(ans))
```

# Exercise 2.8

We will use the same method as Newton did for calculating an approximation for $\pi$. By calculating cleverly a subpart of the unitary disk area,
Newton obtains the following formula:

$$
\begin{align*}
\frac{\pi} {12} = \int _{0} ^{ \frac{1} {2}} \sqrt {1-x^2} dx - \frac{\sqrt{3}} {8}
\end{align*}
$$

* Check symbolically that this formula is valid.

```python
sym.integrate( sym.sqrt(1-x**2), (x,0,sym.Rational(1,2))) - sym.sqrt(3) / 8
```

Newton used Taylor approximation to calculate the first decimals of $\pi$.
 * Given an order n = 3, Calculate a Taylor series $s_1$ at 0 of order n that approximates the square root under the integral.
 * Deduce a Taylor series $s_2$ for the integral (Hint: don't consider the bounds)
 * Remove the error/remainder by applying the removeO() function for obtaining $s_3$
 * Calculate the integral I by substituting the bound in $s_3$
 * Deduce an approximation for $\pi$
 * Calculate the error s as a difference float between sym.pi and the previous approximation.

```python
n = 3
s1 = sym.series( sym.sqrt(1-x**2), x, 0, n)
s1
```

```python
s2 = sym.integrate(s1, x)
s2
```

```python
s3 = s2.removeO()
s3
```

```python
I = s3.subs({x:sym.Rational(1,2)})
I
```

```python
pi = 12 * I - 12 * sym.sqrt(3)/8
pi, pi.evalf()
```

```python
error = pi - sym.pi
error.evalf()
```

* By collecting the previous statements, code a program that increases the order of the Taylor series up to a given precision for PI approximation is reached.

```python
precision = 10E-14
error = 1
n = 3
while (error > precision):
    s1 = sym.series( sym.sqrt(1-x**2), x, 0, n)
    s2 = sym.integrate(s1, x)
    s3 = s2.removeO()
    I = s3.subs({x:sym.Rational(1,2)})
    pi = 12 * I - 12 * sym.sqrt(3)/8
    error = (pi - sym.pi).evalf()
    print (pi, error, n)
    n += 1
pi.evalf()
```
