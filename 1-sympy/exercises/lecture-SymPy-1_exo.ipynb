We will experiment with the `solve` and `solveset` functions to find polynomial roots or solve equations.

* Find the roots of the equation $x^2+ax+1=0$.
* Find the solutions by substituting $a=1$ and $a=2\sqrt{2}$.

```python
p1 = x**2 + a*x + 1
sym.solve(p1,x)
```

```python
roots = sym.solve(p1,x)
[ r.subs({a:1}) for r in roots ]
```

* Find the roots of the following equation specified using an `Eq` instruction: $x^3 = p*x + q$
* The form of the resulting solution is very similar to Cardano's solution, expanded in the 16th century.

```python
p,q = sym.symbols("p q")
eq1 = sym.Eq(x**3, p*x + q)
roots = sym.solve(eq1,x)
roots[0].simplify()
```

```python
roots[1].factor()
```

```python
roots[2].factor()
```

```python
roots[0].subs({p:15, q:4})
```
